# Helicity
Added 4/29/25 with updated scores (threshold of 100 reads, averaged across BCs, error = standard deviation). 

## Overview of helical predictions within Sog1: 
- Steps:
    - I used the "dssp_helicity()" function from Sparrow to predict the helicity of the wild type Sog1 Sequence.
    - In the top panel below, the trace is the per residue probability of helicity. The pink rectangles behind represnt areas which the function calls as helical.
    - In the bottom panel, I ran the dssp_helicity() function on each wild type tile. I averaged the probabiltiies to get one score per tile, which is the pink trace. The blue trace is the activity measured at each center.
- Takeaway: The N-terminal AD and first C-terminal AD both overlap predicted helical areas. This is consistent with previously annotated helices. 

```python
display(Image(filename="../output/figures/helicity/helicity_overview.png", width=800))
```
