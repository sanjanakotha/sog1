```python
# Threshold applied to BCs
thresholds = []
unique_ADs = []

for thresh in np.arange(0, 1000, 5):
    rows_keep = bc_activ_reads[bc_activ_reads["reads"] > thresh]
    thresholds.append(thresh)
    unique_ADs.append(len(rows_keep["AD"].unique()))

threshold_AD_loss = pd.DataFrame({"threshold" : thresholds, "unique_ADs" : unique_ADs})
sns.lineplot(data = threshold_AD_loss, x = "threshold", y = "unique_ADs")
plt.xlabel("BC Reads Threshold")
plt.ylabel("ADs Recovered")
plt.axvline(100, color = 'red')
sns.despine()
```

# Variability between two sequencing runs?

```python
both_scores = pd.read_csv("../output/Sog1_library2_activities_with_reads_SK.csv")
both_scores
```

```python
sort_2_scores = pd.read_csv("../data/Sog1_library1_sort2.csv")
sort_2_scores
```

```python
S2_S3_scores = pd.merge(both_scores, sort_2_scores, on = "AAseq", how = "left")
S2_S3_scores
```

```python
sns.pairplot(data = S2_S3_scores[["Activity_S3_1", "Activity_S3_2", "Activity_GFP8"]])
```

```python
sns.scatterplot(data = S2_S3_scores, x = "Activity_S3_1", y = "Activity_S3_2", s = 3, alpha = 0.1)
```

```python
df = S2_S3_scores.dropna(subset = ["Activity_S3_1", "Activity_S3_2"])
x = df["Activity_S3_1"]
y = df["Activity_S3_2"]

m, b = np.polyfit(x, y, 1)
m, b
```

```python
S2_S3_scores["exp_Activity_S3_2"] = m * S2_S3_scores["Activity_S3_1"] + b
S2_S3_scores
```

```python
sns.scatterplot(data = S2_S3_scores, x = "Activity_S3_1", y = "Activity_GFP8", s = 3, alpha = 0.5)
df = S2_S3_scores.dropna(subset = ["Activity_S3_1", "Activity_GFP8"])
x = df["Activity_S3_1"]
y = df["Activity_GFP8"]

m, b = np.polyfit(x, y, 1)
m, b
```

```python
S2_S3_scores["exp_Activity_GFP8"] = m * S2_S3_scores["Activity_GFP8"] + b
S2_S3_scores
```

```python
S2_S3_scores[["Activity_S3_1", "Activity_GFP8", "exp_Activity_GFP8"]].dropna()
```

```python
S2_S3_scores["reg_S3_1_2_std"] = np.std(S2_S3_scores[["Activity_S3_2", "exp_Activity_S3_2"]], axis = 1)
S2_S3_scores["reg_S3_1_S2_std"] = np.std(S2_S3_scores[["Activity_GFP8", "exp_Activity_GFP8"]], axis = 1)
S2_S3_scores
```
