```python
mys_path = "/global/scratch/projects/fc_mvslab/data/sequencing/bys4/first_sequencing/MYS2/"

datafiles_gfp8 = ["GTAC-125749-MYS_1.HYKNTDSX3_AACGCAATTC-CGTATCCGTA_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_2.HYKNTDSX3_CGTCGCTAAG-AATACCAGGA_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_3.HYKNTDSX3_TGGACTCGAA-TTCCGTATTG_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_4.HYKNTDSX3_TTAGGTTGTC-TCTTCAACCG_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_5.HYKNTDSX3_GCCATTCATA-TAGTGTACTG_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_6.HYKNTDSX3_TCCAGCCTAA-AGCGCGGATT_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_7.HYKNTDSX3_GCTTGATATC-TGCACCTTAT_L002.fastq.gz.assembled.fastq.gz",
            "GTAC-125749-MYS_8.HYKNTDSX3_GTCGCACAGA-CGAGTTGGAT_L002.fastq.gz.assembled.fastq.gz"]
datafiles_gfp8 = [mys_path + filepath for filepath in datafiles_gfp8]

datafiles_gfp8.reverse()
bin_counts = [100,100,100,100,100,100,100,100]
bin_values_gfp8 = [159,258,330,417,559,856,1804,8193]
gfp8wayfiles = Sort(datafiles_gfp8, bin_counts, bin_values_gfp8, bc_dict = True)
```

```python
gfp8_output = gfp8wayfiles.process(thresh = 0)
len(gfp8_output)
```

```python
gfp_activity_bc, gfp_reads_bc, gfp_bins_bc, gfp_loss_table_bc  = gfp8_output
gfp_activity_bc
```

```python
ad_bcs = seq_counter(datafiles_gfp8, design_to_use="../data/unique_seqs.csv", barcoded = True)
ad_bcs
```

```python
from labtools.adtools.counter import create_map
from labtools.adtools.seqlib import write_bc_dict

bc_map = create_map(ad_bcs[0])
bc_map
```

```python
bc_map_df = pd.DataFrame({"BC" : bc_map[0].keys(), "AD" : bc_map[0].values()})
bc_map_df
```

```python
gfp_activity_bc_to_merge = gfp_activity_bc.reset_index()
gfp_activity_bc_to_merge = gfp_activity_bc_to_merge.rename(columns = {"index" : "BC"})
gfp_activity_bc_to_merge
```

```python
gfp_activity_bc_with_ad = pd.merge(gfp_activity_bc_to_merge, bc_map_df)
gfp_activity_bc_with_ad
```

```python
gfp_activity_bc_with_ad.to_csv("../output/gfp_activity_bc_with_ad.csv")
```

# 2. Analyzing

```python
gfp_activity_bc_with_ad = pd.read_csv("../output/gfp_activity_bc_with_ad.csv", index_col = 0)
gfp_activity_bc_with_ad
```

```python
reads = pd.DataFrame(gfp_reads_bc)
reads = reads.reset_index()
reads = reads.rename(columns = {"index" : "BC", 0 : 'reads'})
gfp_activity_bc_with_ad = pd.merge(gfp_activity_bc_with_ad, reads, on = "BC")
gfp_activity_bc_with_ad
```

```python
gfp_activity_bc_with_ad.to_csv("../output/gfp_activity_bc_with_ad_reads.csv")
```

```python
gfp_activity_bc_with_ad[gfp_activity_bc_with_ad["reads"] == 1]
```

```python
# How many reads per BC? - dropping singletons
sns.set_context('talk')
ax = sns.histplot(gfp_activity_bc_with_ad["reads"], binwidth = 10)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel("Sort2: Reads per BC")
ax.axvline(25, color = 'red')
sns.despine()
```

```python
# How many ADs recovered at each threshold?
```

```python
sns.set_context('talk')
thresh = []
AD_count = []
for i in np.arange(0, 500, 10):
    thresh.append(i)
    AD_count.append(len(gfp_activity_bc_with_ad[gfp_activity_bc_with_ad["reads"] > i]["AD"].unique()))
sns.lineplot(x = thresh, y = AD_count)
plt.xlabel("Sort2: BC Reads Threshold")
plt.ylabel("ADs Recovered")
plt.axvline(25, color = 'red')
sns.despine()
```

```python
gfp_activity_bc_with_ad_over_25 = gfp_activity_bc_with_ad[gfp_activity_bc_with_ad["reads"] > 25]
gfp_activity_bc_with_ad_over_25
```

```python
gfp_activity_bc_with_ad_over_25["AD"].value_counts()
```

```python
ad_stats = gfp_activity_bc_with_ad_over_25[["AD", "Activity"]].groupby("AD").agg(["mean", "std", "count"])
ad_stats = ad_stats.reset_index()
ad_stats.columns = ["AD", "mean", "std", "count"]
ad_stats
```

```python
ad_stats[ad_stats["count"] == 1]
```

```python
sns.set_context('talk')
sns.histplot(ad_stats["count"], binwidth = 1)
plt.xlabel("Sort 2: BCs per AD")
sns.despine()
```

```python
plt.figure(dpi = 300)

ax = sns.scatterplot(data = ad_stats, x = "mean", y = "std", s = 20, alpha = 0.1, edgecolor = 'none')

# Plot y = x line
lims = [
    np.min([ad_stats["mean"].min(), ad_stats["std"].min()]),
    np.max([ad_stats["mean"].max(), ad_stats["std"].max()])
]
ax.plot(lims, lims, '--', color='red', linewidth=1)

# # Optional: set limits to match
# ax.set_xlim(lims)
# ax.set_ylim(lims)

sns.despine()
plt.xlabel("Activity")
plt.ylabel("Error")

plt.show()
```
