# 1. Generating test seq

```python
folder_path = '/global/scratch/projects/fc_mvslab/data/sequencing/CZB_Jan2024/MYS3/feb7/'
folder_path
```

```python
# Specify the path to your FASTQ file

for file in datanames_gfp8:
    fpath = folder_path + file
    with open(fpath, 'r') as f:
        beginning_lines = list(itertools.islice(f, 4*10000))
    
    pd.DataFrame([b.strip() for b in beginning_lines]).to_csv("../data/test_fastqs/test_" + file, index = None, header = None)
```

# 2. TESTING

```python
mys3path = "../data/test_fastqs/test_"
mapfpath_all = '../output/sog1_lib1+2map_EC20250403.csv'#whenever we can save the bothlibsmap_fil to csv
```

```python
datafiles_gfp8 = []
for f in datanames_gfp8:
    fpath = mys3path + f
    datafiles_gfp8.append(fpath)
```

```python
# print(datafiles_gfp8)
bin_counts = [100,100,100,100,100,100,100,100]
bin_values_gfp8 = [146,237,331,483,776,1364,2592,6357]
gfp8wayfiles = Sort(datafiles_gfp8, bin_counts, bin_values_gfp8, bc_dict = True)
```

```python
gfp1_activity, gfp1_reads, gfp1_bins, gfp1_loss_table = gfp8wayfiles.process(ad_preceder = "GCTAGC",
                                   bc_preceder = "GGGCCCG",
                                   bc_anteceder = "GGCGCGCC", 
                                   ad_length = 120,
                                   bclength = 11)
```
