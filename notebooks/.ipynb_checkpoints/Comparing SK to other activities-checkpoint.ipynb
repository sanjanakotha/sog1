```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
SK_scores = pd.read_csv("../output/Sog1_library2_activities_with_reads_SK.csv")
SK_scores
```

```python
EC_scores = pd.read_csv("../data/Sog1_library2_activities_with_reads_ECspike.csv")
EC_scores
```

```python
JS_scores = pd.read_csv("../data/Sog1_library2_activities_with_reads.csv")
JS_scores
```

```python
js_sort2 = pd.read_csv("../data/Sog1_all_activities_and_tiles_sort2.csv")
js_sort2 = js_sort2[["AAseq", "Activity_GFP8"]].drop_duplicates().dropna()
js_sort2
```

```python
sort2_SK = pd.merge(js_sort2, SK_scores, suffixes = ("", "_SK"), how = "left")
sort2_SK_JS = pd.merge(sort2_SK, JS_scores, suffixes = ("_SK", "_JS"), on = "AAseq", how = "left")
sort2_SK_JS_EC = pd.merge(sort2_SK_JS, EC_scores, suffixes = ("_JS", "_EC"), on = "AAseq", how = "left")
sort2_SK_JS_EC = sort2_SK_JS_EC[["Activity_GFP8_SK", "Activity_S3_1_SK", "Activity_S3_1_JS", "Activity_S3_1"]]
sort2_SK_JS_EC = sort2_SK_JS_EC.rename(columns = {"Activity_GFP8_SK" : "S2_GFP8", 
                                                  "Activity_S3_1_SK" : "S3_1_SK",
                                                  "Activity_S3_1_JS" : "S3_1_JS",
                                                  "Activity_S3_1" : "S3_1_EC"})
sort2_SK_JS_EC
```

```python
sns.set_context('talk')
plt.figure(dpi = 900)
sns.pairplot(sort2_SK_JS_EC, plot_kws = {"s" : 10, "alpha" : 0.5}, corner = True)
```
